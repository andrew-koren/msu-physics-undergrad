## Introduction

The primary goal of this project was to develop a system which exhibits nucleation. Nucleation is a broad term, referring to the forming and growing of a percipitate, and appears in many forms in nature. In this case, we want to see our entire system complete a nontrivial transition from an unstable or metastable phase to a stable phase, building from a nucleus.

To do this, I defined a 1D binary chain of sites where each state could be in the ground state ($0$) or excited state ($1$), and experimented with various interactions between sites, and I was able to port over a good deal of code I'd already made for an ising model ($\pm 1$ states).

To make any progress on this problem, I had to answer two questions: how will I evolve the system, and how will sites interact? In tandem with our class material at the time, I decided that working in the canonical ensemble would be a good start. By defining the energy of each state of the chain, I could determine their boltzmann factors and thus probability of appearing analytically. I implemented nearest-neighbor interactions with controlled temperature, with an additional energy penalty for the excited state over the ground state.

Since the total number of binary states grows exponentially $2^N$, these plots could be generated by adding up the probability of each state with the same total number of excited states, and appeared to indicate something similar to what I was looking for. The above plot shows a ground state with very high probability, alongside a middle region that is less probable.

The central point bulges upwards due to a much higher multiplicty of states, however the all-ground state has a higher probability due to having much lower energies and larger boltzmann factors. 

These charts are useful, however analyzing the canonical ensemble alone cannot achieve my goal. For one thing, a central assumption is that *any* state can jump to *any* state, so we cannot grow a precipitate. Nucleation requires sites adjacent to a nucleus to be more likely to go to the same state. To observe nucleation, we need to simulate dynamic evolution of the system.

My first attempt at simulation was quite simple: rather than calculating macrostate probabilites, I imagined each individual site as it's own system, with state energies dependent on adjacent system states. With this, I could observe each site's evolution relative to one another
